## tokenizing


```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
# Extract sentences
data_train_label = data_train.iloc[:, 0].astype(str).tolist()
data_train_feture = data_train.iloc[:, 1].astype(str).tolist()
```

```python
tk = Tokenizer(filters='' , lower=True , )
tk.fit_on_texts(data_train_feture)
```

```python
tk.index_word
```

```python
tk.word_counts
```

```python
tweet_sequences = tk.texts_to_sequences(data_train_feture)
```

```python
tweet_sequences
```

```python
pd.DataFrame(tweet_sequences).shape
```

```python
pd.DataFrame(tweet_sequences).head()
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
tweet_sequences = pad_sequences(tweet_sequences , maxlen=57 , padding="post")
pd.DataFrame(tweet_sequences)
```

```python
vocab_tweet_size = len(tk.word_index)+1
vocab_tweet_size
```

## bild model 

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN , Dense
```

```python
num_classes = len(set(data_train_label))
num_classes
```

```python
model = Sequential([
    Embedding(input_dim=vocab_tweet_size , output_dim=128 , input_length=57 ),
    SimpleRNN(64, activation='tanh'),
    Dense(2 , activation='softmax')
])
```

```python
max_len = 57
model.build(input_shape=(None , max_len))
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
import numpy as np
tweet_sequences = np.array(tweet_sequences)
data_train_label = np.array(data_train_label)
```

```python
data_train_label = np.where(data_train_label == 4, 1, data_train_label)
```

```python
history = model.fit(tweet_sequences , data_train_label, batch_size=36 , epochs=20, validation_split=0.2)
```
